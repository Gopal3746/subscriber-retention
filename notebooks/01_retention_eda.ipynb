# Subscription Retention & LTV — EDA
This notebook explores churn patterns by contract, tenure, payment method, and monthly charges. It uses observational language and leaves survival/LTV estimation to the production pipeline.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT/'data/raw/Telco-Customer-Churn.csv')
df['churn'] = (df['Churn']=='Yes').astype(int)
df.shape
```

```python
df.groupby('Contract')['churn'].agg(['count','mean']).sort_values('mean', ascending=False)
```

```python
tenure_bucket = pd.cut(df['tenure'], [-1,6,12,24,48,72], labels=['0-6','7-12','13-24','25-48','49-72'])
df.groupby(tenure_bucket, observed=True)['churn'].agg(['count','mean'])
```

```python
df.groupby('PaymentMethod')['churn'].agg(['count','mean']).sort_values('mean', ascending=False)
```

```python
df.boxplot(column='MonthlyCharges', by='Churn')
plt.suptitle('')
plt.title('Monthly charges by observed churn status')
plt.show()
```

## Sanity checks
Expected patterns to verify before modeling: month-to-month churn should exceed longer contracts; early-tenure churn should exceed mature-customer churn; and no finding should be interpreted causally from this observational snapshot.
